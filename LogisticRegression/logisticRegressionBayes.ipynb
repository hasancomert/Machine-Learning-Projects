# Lojistik Regresyon ve Naive Bayes Karşılaştırması

Bu çalışmada, laboratuvar föyünün önceki adımlarında incelediğimiz **Lojistik Regresyon** ve **Gaussian Naive Bayes** sınıflandırma algoritmaları aynı veri seti üzerinde eğitilerek performansları karşılaştırılmıştır.

## Algoritmaların Temel Farkları
* **Naive Bayes:** Olasılıksaldır. Veri setindeki özelliklerin (sütunların) birbirinden tamamen bağımsız olduğunu varsayar. Bu varsayım gerçek hayatta her zaman doğru olmasa da çok hızlı çalışmasını sağlar.
* **Lojistik Regresyon:** Doğrusal bir modeldir. Özellikler arasındaki ilişkileri (ağırlıkları) öğrenerek bir olasılık değeri (Sigmoid fonksiyonu ile) üretir. Özellikler arası korelasyon olduğunda genelde daha iyi sonuç verir.

```python
# Gerekli kütüphaneleri çağırıyoruz
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# 1. Veri Hazırlığı
# Karşılaştırma yapabilmek için 500 satırlık rastgele bir veri seti oluşturuyoruz
X, y = make_classification(n_samples=500, n_features=4, random_state=42)

# Veriyi %80 eğitim, %20 test olacak şekilde bölüyoruz
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 2. Modelleri Tanımlama
model_lr = LogisticRegression() # Lojistik Regresyon Modeli
model_nb = GaussianNB()         # Naive Bayes Modeli

# 3. Modelleri Eğitme (Öğrenme Aşaması)
model_lr.fit(X_train, y_train)
model_nb.fit(X_train, y_train)

# 4. Test Verisi Üzerinden Tahmin Yaptırma
tahmin_lr = model_lr.predict(X_test)
tahmin_nb = model_nb.predict(X_test)

# 5. Sonuçları Hesaplama ve Ekrana Yazdırma
basari_lr = accuracy_score(y_test, tahmin_lr)
basari_nb = accuracy_score(y_test, tahmin_nb)

print("--- Performans Karşılaştırması ---")
print(f"Lojistik Regresyon Başarısı: % {basari_lr * 100:.2f}")
print(f"Naive Bayes Başarısı       : % {basari_nb * 100:.2f}")
```

## Sonuç
Yukarıdaki test sonucunda iki algoritmanın da aynı veri setine verdiği tepki ölçülmüştür. Veri setinin yapısına, boyutuna ve özellikler (features) arasındaki ilişkilere bağlı olarak hangi algoritmanın daha başarılı olacağı değişiklik gösterebilir. Bu örnekte algoritmaların doğruluk (accuracy) skorları karşılaştırmalı olarak ekrana yazdırılmıştır.
