# Lojistik Regresyon ve Özdeğer Analizi

## 1. Lojistik Regresyon Nedir?
Lojistik regresyon, bir verinin belirli bir sınıfa ait olma olasılığını hesaplayan bir sınıflandırma algoritmasıdır. Çıktıyı 0 ile 1 arasına sıkıştırmak için sigmoid fonksiyonunu kullanır.

## 2. Numpy `linalg.eig` Fonksiyonu İncelemesi
Bu bölümde laboratuvar föyünde istenen `numpy.linalg.eig` fonksiyonunun dokümantasyonu ve kaynak kodları incelenmiştir.

**Fonksiyonun Amacı:** Karesel bir matrisin özdeğerlerini (eigenvalues) ve sağ özvektörlerini (eigenvectors) hesaplar.
**Parametreler:** İşlem yapılacak kare bir matris alır.
**Dönen Değerler:** 1. `w`: Özdeğerleri içeren bir dizi.
2. `v`: İlgili özdeğerlere karşılık gelen normalize edilmiş özvektörleri içeren matris.

**Kaynak Kod Analizi:** GitHub üzerindeki `numpy/linalg` kaynak kodları incelendiğinde, bu işlemin arka planda lineer cebir hesaplamaları için optimize edilmiş LAPACK kütüphanesini (özellikle `geev` rutinlerini) çağırdığı görülmektedir. Bu sayede karmaşık matris işlemleri C ve Fortran hızında gerçekleştirilerek sisteme büyük bir performans avantajı sağlar.

```python
# Gerekli kütüphaneleri ekliyoruz
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# --- 1. KISIM: Özdeğer ve Özvektör Hesaplama ---
print("--- Özdeğer ve Özvektör Analizi ---")

# 2x2 boyutunda karesel bir matris tanımlıyoruz
A = np.array([[4, -2],
              [1,  1]])

# linalg.eig fonksiyonunu çağırıp dönen değerleri değişkenlere atıyoruz
ozdegerler, ozvektorler = np.linalg.eig(A)

# Sonuçları ekrana yazdırıyoruz
print("Kullanılan Matris:")
print(A)
print("\nHesaplanan Özdeğerler:")
print(ozdegerler)
print("\nHesaplanan Özvektörler:")
print(ozvektorler)
print("\n------------------------------------------------\n")

# --- 2. KISIM: Lojistik Regresyon Uygulaması ---
print("--- Lojistik Regresyon Sınıflandırması ---")

# Test için rastgele veriler oluşturuyoruz (100 satır, 2 özellik)
X, y = make_classification(n_samples=100, n_features=2, n_redundant=0, random_state=42)

# Veriyi eğitim (%80) ve test (%20) olarak ikiye ayırıyoruz
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Modeli oluşturup eğitim verileriyle eğitiyoruz
model = LogisticRegression()
model.fit(X_train, y_train)

# Test verileri üzerinden tahmin yaptırıyoruz
tahminler = model.predict(X_test)

# Başarı oranını hesaplayıp ekrana yazdırıyoruz
basari = accuracy_score(y_test, tahminler)
print("Modelin Doğruluk Oranı: %", basari * 100)
```
